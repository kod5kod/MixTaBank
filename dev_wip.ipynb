```python
apartment_rent_df['bathrooms'] = apartment_rent_df['bathrooms'].replace(["No",], 0, inplace=False)
apartment_rent_df['bathrooms'] = apartment_rent_df['bathrooms'].replace(["Thumbnail",], np.nan, inplace=False)
apartment_rent_df['bathrooms'] = apartment_rent_df['bathrooms'].astype("Float64")

apartment_rent_df['bedrooms'] = apartment_rent_df['bedrooms'].replace(["No",], 0, inplace=False)
apartment_rent_df['bedrooms'] = apartment_rent_df['bedrooms'].replace(["Thumbnail","Cats,Dogs"], np.nan, inplace=False)
apartment_rent_df['bedrooms'] = apartment_rent_df['bedrooms'].astype("Float64").astype("Int64")

apartment_rent_df["square_feet"] = apartment_rent_df["square_feet"].replace(["NJ","CA"], np.nan)
apartment_rent_df = apartment_rent_df.astype({"square_feet": "Float64", }).astype({"square_feet": "Int64", })

apartment_rent_df["state"] = apartment_rent_df["state"].replace(["RentDigs.com"], np.nan)
apartment_rent_df = apartment_rent_df.astype({"state": "object", })

cols_to_keep = ['category', 'bathrooms', 'bedrooms',   'amenities',
       'currency', 'fee', 'has_photo', 'pets_allowed', 'price',
       'price_type', 'square_feet',   'cityname',
       'state',  'source',  ]
# removed: {'address', 'body', 'latitude', 'longitude', 'price_display', 'time', 'title'}
apartment_rent_new_df = apartment_rent_df[cols_to_keep]
apartment_rent_new_df
```
